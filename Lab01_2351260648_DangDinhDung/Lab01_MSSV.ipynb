```python
import os
import numpy as np
from pydub import AudioSegment

# 1. Đọc tệp âm thanh
file_path = r"D:\Audio & Speech\Lab01_2351260648_DangDinhDung\audio\input.wav"
audio = AudioSegment.from_file(file_path)

# 2. Trích xuất và kiểm tra metadata
Fs = audio.frame_rate
channels = audio.channels
sample_width = audio.sample_width
duration = len(audio) / 1000.0  # pydub tính chiều dài theo mili-giây
file_size_kb = os.path.getsize(file_path) / 1024

print("=== THÔNG TIN METADATA BƯỚC A ===")
print(f"Tần số lấy mẫu (Fs): {Fs} Hz")
print(f"Số kênh: {channels}")
print(f"Độ rộng mẫu (Sample width): {sample_width} byte(s) ({sample_width * 8} bit)")
print(f"Thời lượng: {duration} giây")
print(f"Kích thước tệp: {file_size_kb:.2f} KB")

# 3. Chuyển về Mono và chuẩn hóa về miền [-1, 1]
mono = audio.set_channels(1)
# Trích xuất mảng giá trị raw của âm thanh
x = np.array(mono.get_array_of_samples(), dtype=np.float64)
# Phép chia chuẩn hóa dựa trên bit depth
x = x / (2**(8 * mono.sample_width - 1))

print("\n=== KẾT QUẢ CHUẨN HÓA ===")
print(f"Peak (Biên độ cực đại): {np.max(np.abs(x)):.4f}")
print(f"RMS (Mức năng lượng hiệu dụng): {np.sqrt(np.mean(x**2)):.4f}")
```
